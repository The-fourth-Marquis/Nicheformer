### liver data preparation

```python
import os
import numpy as np
import pandas as pd
import scipy.io
import anndata as ad
from scipy import sparse


def read_gene_by_cell_mtx(path):
    # R 导出的 MatrixMarket 是 gene × cell
    # AnnData 需要 cell × gene
    return scipy.io.mmread(path).T.tocsr()


def read_reduction(path, obs_names):
    df = pd.read_csv(path, index_col=0)
    df = df.reindex(obs_names)
    return df.to_numpy()


def build_h5ad(folder, output_h5ad):
    # obs / cells / genes
    cells = pd.read_csv(os.path.join(folder, "cells.csv"))["cell"].astype(str).values
    
    obs = pd.read_csv(os.path.join(folder, "metadata.csv"), index_col=0)
    obs.index = obs.index.astype(str)
    obs = obs.reindex(cells)
    
    genes = pd.read_csv(os.path.join(folder, "rna_genes.csv"))["gene"].astype(str).values
    var = pd.DataFrame(index=genes)
    
    # RNA matrix
    counts = read_gene_by_cell_mtx(os.path.join(folder, "rna_counts_gene_by_cell.mtx"))
    data = read_gene_by_cell_mtx(os.path.join(folder, "rna_data_gene_by_cell.mtx"))
    
    # 为了可视化，推荐把 normalized/log data 放在 X
    # raw counts 放在 layers["counts"]
    adata = ad.AnnData(X=data, obs=obs, var=var)
    adata.layers["counts"] = counts
    adata.layers["data"] = data
    
    # spatial coordinates
    if {"x_slide_mm", "y_slide_mm"}.issubset(adata.obs.columns):
        adata.obsm["spatial"] = adata.obs[["x_slide_mm", "y_slide_mm"]].to_numpy()
    
    if {"x_FOV_px", "y_FOV_px"}.issubset(adata.obs.columns):
        adata.obsm["spatial_fov_px"] = adata.obs[["x_FOV_px", "y_FOV_px"]].to_numpy()
    
    # reductions
    red_map = {
        "reduction_pca.csv": "X_pca",
        "reduction_approximateumap.csv": "X_umap",
        "reduction_approximateUMAP_bySlide.csv": "X_umap_bySlide",
    }
    
    for filename, obsm_key in red_map.items():
        path = os.path.join(folder, filename)
        if os.path.exists(path):
            adata.obsm[obsm_key] = read_reduction(path, adata.obs_names)
    
    # falsecode assay
    false_path = os.path.join(folder, "falsecode_counts_feature_by_cell.mtx")
    false_feat_path = os.path.join(folder, "falsecode_features.csv")
    
    if os.path.exists(false_path):
        false_counts = read_gene_by_cell_mtx(false_path)
        false_features = pd.read_csv(false_feat_path)["feature"].astype(str).values
        
        # 注意：falsecode 的 feature 维度不同，不能放进 layers
        # 放到 obsm，并把 feature names 放进 uns
        adata.obsm["falsecode_counts"] = false_counts
        adata.uns["falsecode_features"] = false_features
    
    # negprobes assay
    neg_path = os.path.join(folder, "negprobes_counts_feature_by_cell.mtx")
    neg_feat_path = os.path.join(folder, "negprobes_features.csv")
    
    if os.path.exists(neg_path):
        neg_counts = read_gene_by_cell_mtx(neg_path)
        neg_features = pd.read_csv(neg_feat_path)["feature"].astype(str).values
        
        adata.obsm["negprobes_counts"] = neg_counts
        adata.uns["negprobes_features"] = neg_features
    
    adata.write_h5ad(output_h5ad, compression="gzip")
    return adata

```
